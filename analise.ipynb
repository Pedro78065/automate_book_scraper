```python
import pandas as pd

df = pd.read_csv('produtos.csv', sep=',')
display(df)
display(df.info())
```

```python
df["Price"] = df["Price"].str.extract(r"£(\d+\.\d+)").astype(float)
livro_barato = df[df['Price'] < 20]
```

```python
display(df)
display(livro_barato)
```

```python
df = pd.DataFrame(livro_barato)
df.to_csv('livros_baratos.csv', index=False)
df = pd.read_csv('livros_baratos.csv')
display(df)
```
